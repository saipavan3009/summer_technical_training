```python
'''
1. insertion
2. preorder
3. postorder
4. inorder
5. levelorder
6. search
7. leaf_node_order
8. sum_node_order
9. count_nodes
10. height_of tree
'''
from collections import deque

class TreeNode:
    def __init__(self,data):
        self.left=None
        self.right=None
        self.data=data

    # 1. Insertion
    def insert(self,data):
        if data < self.data:
            if self.left is None:
                self.left = TreeNode(data)
            else:
                self.left.insert(data)
        else:
            if self.right is None:
                self.right = TreeNode(data)
            else:
                self.right.insert(data)

    # 2. Preorder
    def preorder(self):
        print(self.data,end=" ")
        if self.left:
            self.left.preorder()
        if self.right:
            self.right.preorder()

    # 3. Inorder
    def inorder(self):
        if self.left:
            self.left.inorder()

        print(self.data,end=" ")

        if self.right:
            self.right.inorder()

    # 4. Postorder
    def postorder(self):
        if self.left:
            self.left.postorder()

        if self.right:
            self.right.postorder()

        print(self.data,end=" ")

    # 5. Level Order
    def levelorder(self):
        q = deque([self])

        while q:
            node = q.popleft()
            print(node.data,end=" ")

            if node.left:
                q.append(node.left)

            if node.right:
                q.append(node.right)

    # 6. Search
    def search(self,data):
        if self.data == data:
            print("Node Found")
            return

        if data < self.data:
            if self.left:
                self.left.search(data)
            else:
                print("Node Not Found")
        else:
            if self.right:
                self.right.search(data)
            else:
                print("Node Not Found")

    # 7. Leaf Nodes
    def leaf_nodes(self):
        if self.left is None and self.right is None:
            print(self.data,end=" ")

        if self.left:
            self.left.leaf_nodes()

        if self.right:
            self.right.leaf_nodes()

    # 8. Sum of Nodes
    def sum_nodes(self):
        total = self.data

        if self.left:
            total += self.left.sum_nodes()

        if self.right:
            total += self.right.sum_nodes()

        return total

    # 9. Count Nodes
    def count_nodes(self):
        count = 1

        if self.left:
            count += self.left.count_nodes()

        if self.right:
            count += self.right.count_nodes()

        return count

    # 10. Height of Tree
    def height(self):
        left_height = 0
        right_height = 0

        if self.left:
            left_height = self.left.height()

        if self.right:
            right_height = self.right.height()

        return max(left_height,right_height) + 1


# Driver Code
tree = TreeNode(8)

tree.insert(3)
tree.insert(5)
tree.insert(7)
tree.insert(10)
tree.insert(21)
tree.insert(25)
tree.insert(30)
tree.insert(100)

print("Preorder:")
tree.preorder()

print("\n\nInorder:")
tree.inorder()

print("\n\nPostorder:")
tree.postorder()

print("\n\nLevel Order:")
tree.levelorder()

print("\n\nLeaf Nodes:")
tree.leaf_nodes()

print("\n\nSearch:")
tree.search(21)

print("\nSum of Nodes:", tree.sum_nodes())

print("Count of Nodes:", tree.count_nodes())

print("Height of Tree:", tree.height())
```
