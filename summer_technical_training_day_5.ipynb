```python
# ============================================
# MINI PROJECT
# BANK ACCOUNT MANAGEMENT SYSTEM
# Demonstrates all 4 OOP Pillars in Python
# ============================================

# ----------- ABSTRACTION -----------
# Using abstract class to define common structure

from abc import ABC, abstractmethod


class BankAccount(ABC):

    # ----------- ENCAPSULATION -----------
    # balance is made private using __balance

    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.__balance = balance

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method for deposit
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Invalid deposit amount.")

    # Protected method to update balance internally
    def _deduct_balance(self, amount):
        self.__balance -= amount

    # Abstract method (Abstraction)
    @abstractmethod
    def withdraw(self, amount):
        pass

    def display_details(self):
        print("\n------ Account Details ------")
        print("Account Holder :", self.account_holder)
        print("Account Number :", self.account_number)
        print("Balance         :", self.__balance)


# ----------- INHERITANCE -----------
# SavingsAccount inherits BankAccount

class SavingsAccount(BankAccount):

    def __init__(self, account_holder, account_number, balance, interest_rate):
        super().__init__(account_holder, account_number, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.get_balance() * self.interest_rate / 100
        self.deposit(interest)
        print(f"Interest of ₹{interest} added.")

    # ----------- POLYMORPHISM -----------
    # Different withdraw implementation

    def withdraw(self, amount):
        if amount <= self.get_balance():
            self._deduct_balance(amount)
            print(f"₹{amount} withdrawn from Savings Account.")
        else:
            print("Insufficient balance in Savings Account.")


# ----------- INHERITANCE -----------
# CurrentAccount inherits BankAccount

class CurrentAccount(BankAccount):

    def __init__(self, account_holder, account_number, balance, withdrawal_limit):
        super().__init__(account_holder, account_number, balance)
        self.withdrawal_limit = withdrawal_limit

    # ----------- POLYMORPHISM -----------
    # Different withdraw implementation

    def withdraw(self, amount):

        if amount > self.withdrawal_limit:
            print("Withdrawal limit exceeded.")

        elif amount > self.get_balance():
            print("Insufficient balance in Current Account.")

        else:
            self._deduct_balance(amount)
            print(f"₹{amount} withdrawn from Current Account.")


# ============================================
# MAIN PROGRAM
# ============================================

print("======= BANK ACCOUNT MANAGEMENT SYSTEM =======")

# Creating Savings Account Object
savings = SavingsAccount(
    "Sai Pavan",
    1001,
    5000,
    5
)

# Creating Current Account Object
current = CurrentAccount(
    "Rahul",
    2001,
    10000,
    3000
)

# ---------- SAVINGS ACCOUNT OPERATIONS ----------

print("\n===== SAVINGS ACCOUNT =====")

savings.display_details()

savings.deposit(2000)

savings.withdraw(1500)

savings.add_interest()

print("Final Balance :", savings.get_balance())

# ---------- CURRENT ACCOUNT OPERATIONS ----------

print("\n===== CURRENT ACCOUNT =====")

current.display_details()

current.deposit(5000)

current.withdraw(2000)

current.withdraw(5000)   # Exceeds withdrawal limit

print("Final Balance :", current.get_balance())


# ============================================
# OOP PILLARS USED
# ============================================

print("\n\n===== OOP PILLARS DEMONSTRATED =====")

print("1. ENCAPSULATION")
print("   -> Balance is private using __balance")

print("\n2. ABSTRACTION")
print("   -> Abstract class BankAccount with abstract withdraw()")

print("\n3. INHERITANCE")
print("   -> SavingsAccount and CurrentAccount inherit BankAccount")

print("\n4. POLYMORPHISM")
print("   -> withdraw() behaves differently for each account type")
```
