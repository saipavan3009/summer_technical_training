```python
def is_prime(n):
  if n <= 1:
    return False
  for i in range(2,int(n**0.5)+1):
    if n%i==0:
      return False
  return True
def is_perfect_square(n):
  root=int(n**0.5)
  return root*root==n
def generate_series(N):
  series=[]
  power_of_2=0
  power_of_3=0
  for pos in range(1,N+1):
    if is_prime(pos):
      series.append(2**power_of_2)
      power_of_2+=1
    elif is_perfect_square(pos):
      series.append(3**power_of_3)
      power_of_3+=1
    else:
      val=series[-1]+series[-2]
      series.append(val)
  return series
N=19
result=generate_series(N)
print(result[N-1])
```
