```python
'''
you are the head of the election committee in your village.
Each Political party is associated with a unique number and the party number voted for by the villagers.
For a party to win, they must have a majority of votes.
our task is to find and return an integer value denoting the winning party's number.
Return -1 if there is no party with a majority.
Note: if only one vote is there he is the winner.
'''
```

```python
from collections import Counter

def find_winning_party(votes):
    """
    Finds the winning party based on a list of votes.

    Args:
        votes: A list of integers, where each integer represents a vote for a party.

    Returns:
        An integer denoting the winning party's number if a majority exists.
        Returns -1 if there is no party with a majority.
    """
    n = len(votes)
    if n == 0:
        return -1
    if n == 1:
        return votes[0]

    # Count the occurrences of each party vote
    vote_counts = Counter(votes)

    # Calculate the majority threshold (more than n/2 votes)
    majority_threshold = n // 2

    # Check if any party has a majority
    for party_id, count in vote_counts.items():
        if count > majority_threshold:
            return party_id

    return -1
```

### Explanation of the Code:

1.  **`from collections import Counter`**:
    *   We import the `Counter` class from the `collections` module. `Counter` is a convenient way to count hashable objects in an iterable (like a list) and store them in a dictionary-like object where keys are the items and values are their counts.

2.  **`def find_winning_party(votes):`**:
    *   This defines a function named `find_winning_party` that takes one argument: `votes`, which is expected to be a list of party numbers.

3.  **`n = len(votes)`**:
    *   We get the total number of votes. This is crucial for determining the majority threshold.

4.  **Edge Cases (Empty or Single Vote List)**:
    *   **`if n == 0: return -1`**: If there are no votes, no party can win, so we return -1.
    *   **`if n == 1: return votes[0]`**: As per the note, if there's only one vote, that party is the winner. We directly return the party number.

5.  **`vote_counts = Counter(votes)`**:
    *   Here, `Counter` is used to quickly count how many votes each party received. For example, if `votes = [1, 2, 1, 3, 1]`, `vote_counts` will be `{1: 3, 2: 1, 3: 1}`.

6.  **`majority_threshold = n // 2`**:
    *   A party needs *more than* half of the votes to win. For `n` votes, this means they need `(n // 2) + 1` votes if `n` is even, or `((n-1) // 2) + 1` votes if `n` is odd. A simpler way to represent "more than half" is to check if `count > n // 2`.

7.  **`for party_id, count in vote_counts.items():`**:
    *   We iterate through each party and its vote count in the `vote_counts` dictionary.

8.  **`if count > majority_threshold:`**:
    *   Inside the loop, we check if the current party's vote `count` is greater than the `majority_threshold`. If it is, this party has achieved a majority.

9.  **`return party_id`**:
    *   If a party meets the majority condition, its `party_id` is immediately returned as the winner.

10. **`return -1`**:
    *   If the loop finishes without any party meeting the majority condition, it means no party has a majority, so we return -1.

```python
'''
Alice has a string A consisting of lowercase English letters.
Her friend gives her another string S and asks her to modify string A and replace its characters with the characters present in string S.
But, to achieve the above task, Alice must follow the below steps:
1. Choose a character from string S that has the minimum ASCII distance from the ith character in string A
Replace the ith character in string A with the chosen character in choosen character in string S
Your task is to find and return an integer value, representing minimum total ACII distance that is required to modify string A to the characters in string S. Return 0, if all the character in string S are already present in string A.
sample input:
abcd
xyz
sample output:
86
'''
```

```python
def calculate_min_total_ascii_distance(string_a, string_s):
    total_distance = 0

    if not string_s:
        return 0
    if not string_a:
        return 0

    for char_a in string_a:
        min_dist_for_char_a = float('inf')

        for char_s in string_s:
            dist = abs(ord(char_a) - ord(char_s))
            if dist < min_dist_for_char_a:
                min_dist_for_char_a = dist

        total_distance += min_dist_for_char_a

    return total_distance
string_a_sample = "abcd"
string_s_sample = "xyz"
print(f"A: '{string_a_sample}', S: '{string_s_sample}' -> Min Total ASCII Distance: {calculate_min_total_ascii_distance(string_a_sample, string_s_sample)}")
```
