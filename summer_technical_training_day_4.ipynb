```python
'''
calculate the power
f(x,n)
f(2,3) ---> 2 should be multiple 3 times
            2*2*2
            8
write the python code using recurrsion
'''
def power(base, expo):
  if expo==0:
    return 1
  return base*power(base,expo-1)
print(power(2,3))
```

```python
def fac(n):
  if n==0 or n==1:
    return 1
  return n*fac(n-1)
print(fac(5))
```

```python
def display(L,index):
  if len(L)==index:
    return
  display(L,index+1)
  print(L[index],end=" ")
L=[1,2,10,90,100]
index=0
display(L,index)
```

```python
def string(name,index):
  if(index==len(name)):
    return ""
  return name[index]+string(name,index+1)
name="SRU"
index=0
print(string(name,index))
```

```python
def isPlaindrome(s,i,j):
  if i>=j:
    return True
  if s[i]!=s[j]:
    return False
  return isPlaindrome(s,i+1,j-1)
s="madam"
print(isPlaindrome(s,0,len(s)-1))
```

```python
def Linear_search(L,T,index):
  if index==len(L):
    return False
  if L[index]==T:
    return True
  return Linear_search(L,T,index+1)
L=[10,20,30,40,50]
T=30
index=0
print(Linear_search(L,T,index))
```

```python
def Binary_search(L,low,high,key):
  if low>high:
    return False
  mid=(low+high)//2
  if (L[mid]==key):
    return True
  if (L[mid]<key):
    return Binary_search(L,mid+1,high,key)
  else:
    return Binary_search(L,low,mid-1,key)
L=[10,20,30,40,50]
key=50
low=0
high=len(L)-1
print(Binary_search(L,low,high,key))
```

```python
#back_tracking
def fun(path,n):
  if n==0:
    print(path)
    return
  path.append(n)
  fun(path,n-1)
  path.pop()
  fun(path,n-1)
fun([],3)
```
